# feature engineering 
we are going to add a new column price per bed in the dataset



```python
#['price per bed']
# creating a new column
df['price per bed']=df['price']/df['beds']
df.head()
```

```python
# now groupby price and price per bed (avg price per bed
df.groupby(by='neighbourhood_group')['price per bed'].mean() 
```

```python
# how the price is dependent on neighbourhood
sns.barplot(data=df,x='neighbourhood_group',y='price', hue='room_type')
```

```python
# is there any relation bw n. of reviews and price (scatterplot)
plt.figure(figsize=(10,8))
sns.scatterplot(data=df,x='number_of_reviews',y='price', hue='neighbourhood_group')
plt.title('locality and review analysis')
# this graph shows if price is high chances of review is less.most of the reviews are coming around $600 for queens and brooklyn we are getting max reviews.
```
