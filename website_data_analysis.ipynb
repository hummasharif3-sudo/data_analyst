```python
# q1.what patterns or trends can you observe in website sessions and users over time?sessions and users overtime
sns.set(style="whitegrid")
```

the peak time of sessions is from 12hr to 20 hr time 

```python
plt.figure(figsize=(10,5))
df.groupby("DateHour")[["Sessions","Users"]].sum().plot(ax=plt.gca())
plt.title("Sessions and users over time")
plt.xlabel("DateHour")
plt.ylabel("count")
plt.show()
```

```python
# q2 which marketing channel brought the highest number of users and how we can use it to increase traffic?total users by channel
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Channel Group",y="Users",estimator=np.sum,palette="viridis")
plt.title("total users by channel")
plt.xticks(rotation=45)
plt.show()
```

most users are coming from organic social channel.we need to imporve unassigned,organic video and email channels to increase traffic.

```python
#q 3 which channel has the highest engagement time,and what it tells about users behavior and content effectiveness (avg enegagment time by channel)
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Channel Group",y="Avg engagment",estimator=np.mean,palette="magma")
plt.title("avg engagement  by channel")
plt.xticks(rotation=45)
plt.show()
```

organic video channel has the highest engagment time per session.it engage a user for up to 3 mins on average. email channel is second so it has opportunity to help in growth and it shows viewers are more prone to videos  

```python
# q4 how does the engagment rate vary accross the different traffic channels?(enegagment rate distribution by channel)
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Channel Group",y="Engagement rate",palette="coolwarm")
plt.title("Engagment rate by channel")
plt.xticks(rotation=45)
plt.show()
```

referral channel is giving the highest engagment rate

```python
#q5 which channel is driving more engaged sessions compared to non engaged ones and what should be stratgies to increase(engaged vs non engaged users)
# Group data
session_df = df.groupby("Channel Group")[["Sessions", "Engaged Sessions"]].sum().reset_index()

# Create non-engaged sessions column
session_df["non_engaged"] = (
    session_df["Sessions"] - session_df["Engaged Sessions"]
)

# Melt dataframe
session_df_melted = session_df.melt(
    id_vars="Channel Group",
    value_vars=["Engaged Sessions", "non_engaged"]
)

# Plot
plt.figure(figsize=(8,5))

sns.barplot(
    data=session_df_melted,
    x="Channel Group",
    y="value",
    hue="variable"
)

plt.title("Engaged vs Non-Engaged Sessions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

organic social has most engaged and non engaged sessions.we have to work on direct channel to increase engagment

```python
#q6  at what hour of the day each channel drive the most(traffic by hour and channel)
heatmap_data= df.groupby(["Hour","Channel Group"])["Sessions"].sum().unstack().fillna(0)

plt.figure(figsize=(12,6))

sns.heatmap(
    data=heatmap_data,
    cmap="YlGnBu",
    linewidth=.5,
    annot=True,
    fmt=".0f"
)

plt.title("Traffic by Hour and Channel")
plt.xlabel("Channel Group")
plt.ylabel("Hour of day")
plt.show()
```

on organic social channel most traffic was at night 12 am
on referral channel morning time 11 am was the busy time.

```python
#q7 is there any corelation bw high traffic and high engagement rate over time (engagement rate vs sessions)
# Group data
df_plot = df.groupby("DateHour")[["Engagement rate", "Sessions"]].mean().reset_index()




# Plot
plt.figure(figsize=(10,5))

plt.plot(df_plot["DateHour"], df_plot["Engagement rate"],label="Engagement rate", color="green")
plt.plot(df_plot["DateHour"], df_plot["Sessions"], label="Sessions", color="blue")


plt.title("Engagment rate vs sessions ")
plt.xlabel("DateHour")
plt.legend()
plt.grid(True)

plt.show()
```

engagment rate during the period of one month is zero.there are so many sessions but engagment rate is too low
